```python
import pandas as pd
import numpy as np
```

# 1. Привести различные способы создания объектов типа Series
# Для создания Series можно использовать
# - списки Python или массивы NumPy
# - скалярные значение
# - словари

```python
arr = [1, 2, 3, 4, 5]
first_1 = pd.Series(arr)
first_2 = pd.Series(np.array(arr))

second = pd.Series(1)


dict_th = {'A' : 1, 'B' : 2, 'C' : 3, 'D' : 4, 'E' : 5}
third = pd.Series(dict_th)
```

```python
print(first_1)
print(first_2)
print(second)
print(third)
```

# 2. Привести различные способы создания объектов типа DataFrame
# DataFrame. Способы создания
# - через объекты Series
# - списки словарей
# - словари объектов Series
# - двумерный массив NumPy
# - структурированный массив Numpy

```python
#1
first_df = pd.DataFrame({
    'first_1': first_1, 
    'first_2': first_2, 
})

#2
data = [
    {"Column1": 1, "Column2": 4},
    {"Column1": 2, "Column2": 5},
    {"Column1": 3, "Column2": 6},
]
second_df = pd.DataFrame(data)

#3
series1 = pd.Series([1, 2, 3], name="Column1")
series2 = pd.Series([4, 5, 6], name="Column2")

third_df = pd.DataFrame({series1.name: series1, series2.name: series2})

#4
array = np.array([[1, 4], [2, 5], [3, 6]])
four_df = pd.DataFrame(array, columns=["Column1", "Column2"])

#5
structured_array = np.array(
    [(1, 4), (2, 5), (3, 6)],
    dtype=[("Column1", "int"), ("Column2", "int")]
)

five_df = pd.DataFrame(structured_array)
```

```python
print(first_df)
print()
print(second_df)
print()
print(third_df)
print()
print(four_df)
print()
print(five_df)
```

# 3. Объедините два объекта Series с неодинаковыми множествами ключей (индексов) так, чтобы вместо NaN было установлено значение 1

```python
series1 = pd.Series([10, 20, 30], index=["a", "b", "c"])
series2 = pd.Series([40, 50], index=["b", "d"])

series1 = pd.Series([10, 20, 30], index=["a", "b", "c"])
series2 = pd.Series([40, 50], index=["b", "d"])

result = series1.add(series2, fill_value=0).fillna(1)
```

```python
print(result)
```

# 5. На примере объектов DataFrame продемонстрируйте использование методов ffill() и bfill()

```python
data = {
    "A": [1, np.nan, 3, np.nan],
    "B": [np.nan, 2, np.nan, 4],
    "C": [1, 2, np.nan, np.nan],
}
df = pd.DataFrame(data)
print(df)

df_ffill = df.ffill()
print("\nffill():")
print(df_ffill)

df_bfill = df.bfill()
print("\nbfill():")
print(df_bfill)
```
