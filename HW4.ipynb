# 1. Разобраться как использовать мультииндексные ключи в данном примере

```python
import numpy as np
import pandas as pd
```

```python
index = [
    ('city_1', 2010),
    ('city_1', 2020),
    ('city_2', 2010),
    ('city_2', 2020),
    ('city_3', 2010),
    ('city_3', 2020),
]

population = [
    101,
    201,
    102,
    202,
    103,
    203,
]

pop = pd.Series(population, index = index)
pop_df = pd.DataFrame(
    {
        'total': pop,
        'something': [
            10,
            11,
            12,
            13,
            14,
            15,
        ]
    }
)

i = pd.MultiIndex.from_tuples(index)
pop_df = pop_df.reindex(i)
pop_df
```

```python
pop_1 = pop_df.loc['city_2', 'total']
pop_1
```

```python
pop_1 = pop_df.loc[['city_2', 'city_3'], ['total', 'something']]
pop_1
```

```python
pop_1 = pop_df.loc[['city_1', 'city_2'], 'something']
pop_1
```

# 2. Из получившихся данных выбрать данные по
# - 2020 году (для всех столбцов)
# - job_1 (для всех строк)
# - для city_1 и job_2

```python
index = pd.MultiIndex.from_product(
    [
        ['city_1', 'city_2'],
        ['2010', '2020']
    ],
    names=['city', 'year']
)

columns = pd.MultiIndex.from_product(
    [
        ['person_1', 'person_2', 'person_3'],
        ['job_1', 'job_2']
    ],
    names=['worker', 'job']
)

rng = np.random.default_rng(1)
data_r = rng.random((4, 6))

data = pd.DataFrame(data_r, index=index, columns=columns)
data
```

```python
data_1 = data.xs('2020', level='year')
data_1
```

```python
data_1 = data.xs('job_1', level='job', axis=1)
data_1
```

```python
data_city = data.xs('city_1', level='city')
data_1 = data_city.xs('job_2', level='job', axis=1)
data_1
```

# 4. Привести пример использования inner и outer джойнов для Series

```python
ser1 = pd.Series(['1', '2', '3'], index=[1, 2, 3])
ser2 = pd.Series(['1', '2', '3'], index=[1, 0, 3])

outer_join = pd.concat([ser1, ser2], axis=1, join="outer")
print(outer_join)

inner_join = pd.concat([ser1, ser2], axis=1, join="inner")
print(inner_join)
```
